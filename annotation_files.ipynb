```python
import pandas as pd
import json
import copy
```

```python
fracas_df = pd.read_csv("fracas.csv")
fracas_df_filtered = fracas_df[(fracas_df['multiple_p']==0) & fracas_df['fracas_answer'].isin(['yes', 'no'])]
fracas_df_filtered.head()
```

```python
template = {
    "meta": {
        "sent_id": "",
        "text": "",
        "type": "",
        "fracas_answer": ""
    },
    "s": [],
    "v": [],
    "f": {},
    "e": {},
    "l": {},
    "d": {},
    "c": {},
    "h": {},
    "i": {},
    "labels": {}
    }

for id, fracas_answer, p, h in zip(fracas_df_filtered['problem_id'], 
                                   fracas_df_filtered['fracas_answer'], 
                                   fracas_df_filtered['p'], 
                                   fracas_df_filtered['h']):
    
    
    premise = copy.deepcopy(template)
    premise['meta']['sent_id'] = str(id) + "p"
    premise['meta']['text'] = p
    premise['meta']['type'] = "premise"
    premise['meta']['fracas_answer'] = fracas_answer

    with open(f"annotations/FRACAS/{id}p.yarn.json", "w") as f:
        json.dump(premise, f)

    hypothesis = copy.deepcopy(template)
    hypothesis['meta']['sent_id'] = str(id) + "h"
    hypothesis['meta']['text'] = h
    hypothesis['meta']['type'] = "hypothesis"
    hypothesis['meta']['fracas_answer'] = fracas_answer

    with open(f"annotations/FRACAS/{id}h.yarn.json", "w") as f:
        json.dump(hypothesis, f)
```
